### 2. Escribe un programa Python para contar el número de caracteres (frecuencia de caracteres) en una cadena. Cadena de ejemplo: google.com

```python
def char_frequency(str1):
    dict = {}
    for n in str1:
        keys = dict.keys()
        if n in keys:
            dict[n] += 1
        else:
            dict[n] = 1
    return dict
print(char_frequency('google.com'))
```

### 3. Escribe un programa Python para obtener una cadena hecha de los primeros 2 y los últimos 2 caracteres de una cadena dada. Si la longitud de la cadena es inferior a 2, devuelva en lugar de la cadena vacía. 
Cadena de ejemplo: 'w3resource'

Resultado esperado: 'w3ce'

Cadena de muestra: 'w3'

Resultado esperado: 'w3w3'

Cadena de muestra: ' w'

Resultado esperado: Cadena vacía

```python
def string_both_ends(str):
  if len(str) < 2:
    return ''

  return str[0:2] + str[-2:]

print(string_both_ends('w3resource'))
print(string_both_ends('w3'))
print(string_both_ends('w'))
```

### 4. Escribe un programa de Python para obtener una cadena de una cadena determinada en la que todas las ocurrencias de su primer carácter se han cambiado a $, excepto el primer carácter en sí.

Cadena de ejemplo: 'restart'

Resultado esperado: 'resta$t'

```python
def change_char(str1):
  char = str1[0]
  str1 = str1.replace(char, '$')
  str1 = char + str1[1:]

  return str1

print(change_char('restart'))
```

### 6. Escribe un programa Python para añadir "ing" al final de una cadena determinada (la longitud debe ser de al menos 3). Si la cadena dada ya termina con 'ing', añade 'ly' en su lugar. Si la longitud de la cadena dada es inferior a 3, déjela sin cambios.  

Cadena de ejemplo: 'abc'

Resultado esperado: 'abcing'

Cadena de ejemplo: 'cadena'

Resultado esperado: "stringly"

```python
def add_string(str1):
  length = len(str1)

  if length > 2:
    if str1[-3:] == 'ing':
      str1 += 'ly'
    else:
      str1 += 'ing'

  return str1
print(add_string('ab'))
print(add_string('abc'))
print(add_string('string'))
```

### 7. Escribe un programa de Python para encontrar la primera aparición de la subcadena "no" y "pobre" de una cadena dada, si "no" sigue a la "pobre", reemplaza toda la subcadena "no"... "pobre" por "bueno". Devuelve la cadena resultante. 

Cadena de muestra: "¡La letra no es tan pobre!"
    
"¡La letra es pobre!"

Resultado esperado: "¡La letra es buena!"
    
"¡La letra es pobre!"

```python
def not_poor(str1):
  snot = str1.find('not')
  spoor = str1.find('poor')
  

  if spoor > snot and snot>0 and spoor>0:
    str1 = str1.replace(str1[snot:(spoor+4)], 'good')
    return str1
  else:
    return str1
print(not_poor('The lyrics is not that poor!'))
print(not_poor('The lyrics is poor!'))
```

### 8. Escribe una función de Python que tome una lista de palabras y devuelva la palabra más larga y la longitud de la más larga. Ir al editor 

Salida de muestra:

La palabra más larga: Ejercicios

Longitud de la palabra más larga: 9

```python
def find_longest_word(words_list):
    word_len = []
    for n in words_list:
        word_len.append((len(n), n))
    word_len.sort()
    return word_len[-1][0], word_len[-1][1]
result = find_longest_word(["PHP", "Exercises", "Backend"])
print("\nLongest word: ",result[1])
print("Length of the longest word: ",result[0])
```

### 1. Escribe un programa Python para sumar todos los elementos de una lista.

```python
def sum_list(items):
    sum_numbers = 0
    for x in items:
        sum_numbers += x
    return sum_numbers
print(sum_list([1,2,-8]))
```

### 2.  Escribe un programa Python para multiplicar todos los elementos de una lista.

```python
def multiply_list(items):
    tot = 1
    for x in items:
        tot *= x
    return tot
print(multiply_list([1,2,-8]))
```

### 3. Escribe un programa Python para obtener el mayor número de una lista.

```python
def max_num_in_list( list ):
    max = list[ 0 ]
    for a in list:
        if a > max:
            max = a
    return max
print(max_num_in_list([1, 2, -8, 0]))
```

### 5.  Escribe un programa Python para contar el número de cadenas en las que la longitud de la cadena es de 2 o más y el primer y el último carácter son los mismos de una lista dada de cadenas. 

Lista De Muestras: ['abc', 'xyz', 'aba', '1221']

Resultado esperado: 2

```python
def match_words(words):
  ctr = 0

  for word in words:
    if len(word) > 1 and word[0] == word[-1]:
      ctr += 1
  return ctr

print(match_words(['abc', 'xyz', 'aba', '1221']))
```


### 6.  Escribe un programa Python para obtener una lista, ordenada en orden creciente por el último elemento de cada tupla de una lista dada de tuplas no vacías.

Resultado Esperado : [(2, 1), (1, 2), (2, 3), (4, 4), (2, 5)]


```python
def last(n): return n[-1]

def sort_list_last(tuples):
  return sorted(tuples, key=last)

print(sort_list_last([(2, 5), (1, 2), (4, 4), (2, 3), (2, 1)]))
```

### 10. Escribe un programa Python para encontrar la lista de palabras que son más largas que n de una lista determinada de palabras. 

```python
def long_words(n, str):
    word_len = []
    txt = str.split(" ")
    for x in txt:
        if len(x) > n:
            word_len.append(x)
    return word_len	
print(long_words(3, "The quick brown fox jumps over the lazy dog"))
```

### 1. Escribe un script de Python para ordenar (ascendente y descendente) un diccionario por valor.

```python
import operator
d = {1: 2, 3: 4, 4: 3, 2: 1, 0: 0}
print('Original dictionary : ',d)
sorted_d = sorted(d.items(), key=operator.itemgetter(1))
print('Dictionary in ascending order by value : ',sorted_d)
sorted_d = dict( sorted(d.items(), key=operator.itemgetter(1),reverse=True))
print('Dictionary in descending order by value : ',sorted_d)
```
