# ¿Qué es Python?

Python es un lenguaje de alto nivel, de propósito general, multiparadigma, principalmente imperativo, orientado a objetos y funcional, de tipado dinámico y fuertemente tipado a nivel de lenguaje de programación.

La gramática de Python es una de las características más representativas, dado que hace que el lenguaje posea una sintaxis sencilla, simple y clara que permite al programador desarrollar programas de forma intuitiva.


Es posible que hayas oído que **C** es una “lenguaje de nivel inferior.” En el contexto de la programación, la “nivel” es una descripción de lo cerca que está del conjunto de instrucciones nativas de la computadora. Cuanto más bajo sea el nivel, más cerca estás de escribir el código de la máquina. Cuanto más **alto sea el nivel**, más abstracción se hace con el lenguaje para alejar la escritura del código de máquina.

# Entorno virtual
## ¿Por qué necesito Entornos Virtuales?

Las aplicaciones de Python a menudo usan paquetes y módulos que no vienen como parte de la biblioteca estándar. En la mayoría de ocasiones, las aplicaciones necesitarán una versión específica de una librería, ya que la aplicación puede requerir que se corrija un error en particular o que la aplicación se escriba usando una versión obsoleta de la interfaz de la biblioteca.

Esto significa que puede que no sea posible que una instalación de Python cumpla con los requisitos de cada aplicación. Si la aplicación A necesita la versión 1.0 de un módulo en particular, pero la aplicación B necesita la versión 2.0, entonces los requisitos están en conflicto y la instalación de la versión 1.0 o 2.0 dejará una aplicación incapaz de ejecutarse.

La solución para este problema es crear un entorno virtual, un árbol de directorios autocontenido que contenga una instalación de Python para una versión particular.

### Instalación en Linux y Mac:

En el caso de `Linux` y `Mac` viene instalado por defecto una versión de python que puedes verificar con: 

`python3 --version`. 

En caso de no estar instalado puedes descargar directamente desde [https://www.python.org/downloads/](https://www.python.org/downloads/).

#### PASOS:

- Abrir consola
- Crear carpeta en donde será instalado el entorno virtual, elemplo: `mkdir envs/myentorno`
- Crear entorno virtual: `python3 -m venv nombre_entorno`
- Si queremos crear un entorno con una versión específica de Python hacemos: `python3 -m venv nombre_entorno python=3.4`
- Para activar el entorno e instalar librerías: `source envs/myentorno/bin/activate`
- Para desactivar el entorno: `deactivate`.


### Instalación en Windows:

En caso de no estar instalado puedes descargar directamente desde [https://www.python.org/downloads/](https://www.python.org/downloads/).

#### PASOS:

- Abrir consola
- Crear carpeta en donde será instalado el entorno virtual, elemplo: `mkdir envs\myentorno`
- Crear entorno virtual: `python -m venv nombre_entorno`
- Si queremos crear un entorno con una versión específica de Python hacemos: `python -m venv nombre_entorno python=3.4`
- Para activar el entorno e instalar librerías: `envs\nombre_entorno\Scripts\activate.bat`
- Para desactivar el entorno: `deactivate`.
